```python
try:
    with engine.connect() as connection:
        result = connection.execute(text("SELECT current_database();"))
        print("Connection successful")
        print("Connected database:", result.fetchone()[0])
except Exception as e:
    print("Error connecting to the database")
    print(e)
```

```python
create_tables_query = """
CREATE TABLE IF NOT EXISTS agents (
    agent_id VARCHAR(10) PRIMARY KEY,
    name VARCHAR(100),
    phone VARCHAR(30),
    email VARCHAR(100),
    commission_rate NUMERIC(5,2),
    deals_closed INT,
    rating NUMERIC(3,2),
    experience_years INT,
    avg_closing_days INT
);

CREATE TABLE IF NOT EXISTS listings (
    listing_id VARCHAR(10) PRIMARY KEY,
    city VARCHAR(100),
    property_type VARCHAR(50),
    price NUMERIC(15,2),
    sqft NUMERIC(12,2),
    date_listed DATE,
    agent_id VARCHAR(10),
    latitude NUMERIC(10,6),
    longitude NUMERIC(10,6),
    CONSTRAINT fk_listings_agent
        FOREIGN KEY (agent_id) REFERENCES agents(agent_id)
);

CREATE TABLE IF NOT EXISTS property_attributes (
    attribute_id INT PRIMARY KEY,
    listing_id VARCHAR(10) UNIQUE,
    bedrooms INT,
    bathrooms INT,
    floor_number INT,
    total_floors INT,
    year_built INT,
    is_rented BOOLEAN,
    tenant_count INT,
    furnishing_status VARCHAR(50),
    metro_distance_km NUMERIC(6,2),
    parking_available BOOLEAN,
    power_backup BOOLEAN,
    CONSTRAINT fk_attributes_listing
        FOREIGN KEY (listing_id) REFERENCES listings(listing_id)
);

CREATE TABLE IF NOT EXISTS sales (
    listing_id VARCHAR(10) PRIMARY KEY,
    sale_price NUMERIC(15,2),
    date_sold DATE,
    days_on_market NUMERIC(10,2),
    CONSTRAINT fk_sales_listing
        FOREIGN KEY (listing_id) REFERENCES listings(listing_id)
);

CREATE TABLE IF NOT EXISTS buyers (
    buyer_id INT PRIMARY KEY,
    sale_id VARCHAR(10),
    buyer_type VARCHAR(50),
    payment_mode VARCHAR(50),
    loan_taken BOOLEAN,
    loan_provider VARCHAR(100),
    loan_amount NUMERIC(15,2),
    CONSTRAINT fk_buyers_sale
        FOREIGN KEY (sale_id) REFERENCES sales(listing_id)
);
"""
```

```python
with engine.connect() as connection:
    for query in create_tables_query.split(";"):
        if query.strip():
            connection.execute(text(query))
    connection.commit()

print("Tables created successfully")
```

```python
# 7.Percentage of properties sold above listing price
```

```python
query7 = """
SELECT 
    ROUND(
        100.0 * SUM(CASE WHEN s.sale_price > l.price THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS percent_sold_above_listing
FROM listings l
JOIN sales s
ON l.listing_id = s.listing_id;
"""
pd.read_sql(query7, engine)
```

```python
#15. Monthly sales trend
```

```python
query15 = """
SELECT DATE_TRUNC('month', date_sold) AS sale_month,
       COUNT(*) AS total_sales,
       ROUND(SUM(sale_price), 2) AS total_sales_value
FROM sales
GROUP BY sale_month
ORDER BY sale_month;
"""
pd.read_sql(query15, engine)
```

```python
#16. Do properties closer to metro stations command higher prices?
```

```python
query16 = """
SELECT 
    CASE
        WHEN p.metro_distance_km <= 2 THEN '0-2 km'
        WHEN p.metro_distance_km <= 5 THEN '2-5 km'
        WHEN p.metro_distance_km <= 10 THEN '5-10 km'
        ELSE '10+ km'
    END AS metro_distance_bucket,
    ROUND(AVG(l.price), 2) AS avg_price
FROM listings l
JOIN property_attributes p
ON l.listing_id = p.listing_id
GROUP BY metro_distance_bucket
ORDER BY avg_price DESC;
"""
pd.read_sql(query16, engine)
```

```python
#19. Which cities have the highest median property prices?
```

```python
query19 = """
SELECT city,
       ROUND(PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY price)::numeric, 2) AS median_price
FROM listings
GROUP BY city
ORDER BY median_price DESC;
"""
pd.read_sql(query19, engine)
```

```python
#20.How are properties distributed across price buckets?
```

```python
query20 = """
SELECT 
    CASE
        WHEN price < 200000 THEN 'Below 200K'
        WHEN price < 500000 THEN '200K - 500K'
        WHEN price < 1000000 THEN '500K - 1M'
        ELSE 'Above 1M'
    END AS price_bucket,
    COUNT(*) AS property_count
FROM listings
GROUP BY price_bucket
ORDER BY property_count DESC;
"""
pd.read_sql(query20, engine)
```

```python
#23.How does metro distance affect time on market?
```

```python
query23 = """
SELECT 
    CASE
        WHEN p.metro_distance_km <= 2 THEN '0-2 km'
        WHEN p.metro_distance_km <= 5 THEN '2-5 km'
        WHEN p.metro_distance_km <= 10 THEN '5-10 km'
        ELSE '10+ km'
    END AS metro_distance_bucket,
    ROUND(AVG(s.days_on_market), 2) AS avg_days_on_market
FROM property_attributes p
JOIN sales s
ON p.listing_id = s.listing_id
GROUP BY metro_distance_bucket
ORDER BY avg_days_on_market;
"""
pd.read_sql(query23, engine)
```

```python
#29.Which cities have the highest loan uptake rate?
```

```python
query29 = """
SELECT l.city,
       ROUND(
           100.0 * SUM(CASE WHEN b.loan_taken = TRUE THEN 1 ELSE 0 END) / COUNT(*),
           2
       ) AS loan_uptake_rate
FROM buyers b
JOIN sales s
ON b.sale_id = s.listing_id
JOIN listings l
ON s.listing_id = l.listing_id
GROUP BY l.city
ORDER BY loan_uptake_rate DESC;
"""
pd.read_sql(query29, engine)
```
